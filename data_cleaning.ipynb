```python
import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nColumn data types:")
print(df.dtypes)
```

```python
# Keep a copy of the original data
cleaned_df = df.copy()

# Remove duplicate rows
cleaned_df = cleaned_df.drop_duplicates()

# Fill missing numerical values with their median
numeric_cols = cleaned_df.select_dtypes(
    include="number"
).columns

for col in numeric_cols:
    cleaned_df[col] = cleaned_df[col].fillna(
        cleaned_df[col].median()
    )

# Fill missing text values with the mode
text_cols = cleaned_df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_cols:
    if cleaned_df[col].isnull().any():
        mode_values = cleaned_df[col].mode()
        if not mode_values.empty:
            cleaned_df[col] = cleaned_df[col].fillna(
                mode_values.iloc[0]
            )

# Remove unnecessary spaces from text columns
for col in text_cols:
    cleaned_df[col] = cleaned_df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Original rows:", len(df))
print("Cleaned rows:", len(cleaned_df))
print("Missing values after cleaning:")
print(cleaned_df.isnull().sum())
```

```python
cleaned_df.to_csv("cleaned_students_performance.csv",
                  index=False)

print("Cleaned dataset saved successfully!")
```
